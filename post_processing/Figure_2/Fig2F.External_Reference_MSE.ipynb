```r
source("../../R/utils.R")

# Reuse the saved regional-mean comparison; no upstream LD recomputation.
# Source notebook Fig3B_S3D; manuscript Figure 2e retains the three external references.
d <- load_data("Fig3B.LD_Pairwise_Similarity")
refs <- c("ADSP", "UKB", "1000G")
d <- d %>% filter(metric == "mse", x %in% refs, y %in% refs) %>%
  mutate(x = factor(as.character(x), levels = refs),
         y = factor(as.character(y), levels = rev(refs)))
stopifnot(nrow(d) == 9, sum(!is.na(d$val)) == 6)
save_data(d, "Fig2F.External_Reference_MSE")
```
