# Fig 2C — Sketch dimension and LD error

Upper curves retain the constant-population LD-decay planning model and relative matrix MSE criteria. The publication export shows 1–4 cM and adds pairwise LD MSE in the same units as the reference comparison. At B=10,000, the Gaussian Pearson first-order scale 1/B is 0.0001; the observed sketch MSE is 0.0000863. These are precision scales, not fine-mapping error guarantees. See manuscript Methods.


```r
source("../../R/utils.R")

settings <- tidyr::crossing(Ne = c(10000, 20000),
                            relative_mse = c(0.008, 0.010, 0.012)) %>%
  mutate(epsilon = sqrt(relative_mse))
guide <- merge(data.frame(length_cM = seq(0.5, 5, by = 0.025)),
               settings, by = NULL) %>%
  mutate(a = 4 * Ne * length_cM / 100,
         mean_r2 = 2 * ((1 + a) * log1p(a) - a) / a^2,
         rank_proxy = 1 / mean_r2,
         B_plan = ceiling((rank_proxy + 1) / relative_mse)) %>%
  arrange(Ne, relative_mse, length_cM)

save_data(guide, "Fig2E.Sketch_Dimension_Guide")
guide %>% filter(Ne == 10000, length_cM %in% c(1, 2, 3)) %>%
  select(length_cM, Ne, relative_mse, epsilon, B_plan)
```
