# Fig 5E and additional 1000 Genomes stratification — GTEx colocalization, stratified (data processing)

Replication / non-replication with bootstrap 95% CIs over loci, stratified **one way at a time** (pick one for the paper):
- **by CS** — number of SuSiE-RSS CSs at the locus (1, 2, 3+)
- **by z** — quantile of the locus max |z| (4 bins)

Both use the same loci (≥ 1 SuSiE-RSS CS); helpers in `R/utils.R` (`coloc_stratified_summary(by = )`, `coloc_stratified_plot_data`).

**Outputs** (plot-ready list: `plot_df`, `count_lab`, `x_lab`; plus `*_all_methods` tables):
- `figure_data/Fig5C.GTEx_Coloc_by_CS.rds`, `figure_data/Fig5C.GTEx_Coloc_by_Z.rds` — ADSP reference
- `figure_data/GTEx_Coloc_1kG_Strata_by_CS.rds`, `figure_data/GTEx_Coloc_1kG_Strata_by_Z.rds` — 1000 Genomes reference

```r
source("../../R/utils.R")
n_sig_bins <- 4
```

## Fig 5C: ADSP reference

```r
per_record <- read_coloc_records("/sc/arion/projects/ad-omics/haochen/rss_qc/2026_output/Gtex_colocalization/coloc/result")

tb_cs <- coloc_stratified_summary(per_record, by = "cs", B = 1000, seed = 2026)
save_data(tb_cs, "Fig5C.GTEx_Coloc_by_CS_all_methods")
save_data(coloc_stratified_plot_data(tb_cs), "Fig5C.GTEx_Coloc_by_CS")

tb_z <- coloc_stratified_summary(per_record, by = "z", n_sig_bins = n_sig_bins, B = 1000, seed = 2026)
save_data(tb_z, "Fig5C.GTEx_Coloc_by_Z_all_methods")
save_data(coloc_stratified_plot_data(tb_z), "Fig5C.GTEx_Coloc_by_Z")
```

```r
per_record <- read_coloc_records("/sc/arion/projects/ad-omics/haochen/rss_qc/2026_output/Gtex_colocalization_1000G/coloc/result/")

tb_cs <- coloc_stratified_summary(per_record, by = "cs", B = 1000, seed = 2026)
save_data(tb_cs, "GTEx_Coloc_1kG_Strata_by_CS_all_methods")
save_data(coloc_stratified_plot_data(tb_cs), "GTEx_Coloc_1kG_Strata_by_CS")

tb_z <- coloc_stratified_summary(per_record, by = "z", n_sig_bins = n_sig_bins, B = 1000, seed = 2026)
save_data(tb_z, "GTEx_Coloc_1kG_Strata_by_Z_all_methods")
save_data(coloc_stratified_plot_data(tb_z), "GTEx_Coloc_1kG_Strata_by_Z")
```
