# Fig S11E — Effect of the SER fallback on colocalization replication (data processing)

The SER fallback alone does not explain the gain: applying it to SuSiE-RSS (**SuSiE-RSS + SER**) underperforms the full SuSiE-rRSS, and **SuSiE-rRSS -SER** is the EB-mix fit without the fallback. Both are kept only for this supplementary panel; the main Fig 5A/5B show SuSiE-rRSS, SuSiE-RSS and SER.

Reuses the `*_all_methods` tables written by `Fig5A_S11A` and `Fig5B_S11B`, so nothing is re-bootstrapped.

**Output:** `figure_data/FigS11E.Coloc_SER_Fallback_Effect.rds` — list(`GTEx`, `UKB`), ADSP reference, plot-ready

```r
source("../../R/utils.R")
```

```r
## rss+SER          -> SuSiE-RSS + SER
## susie_eb_mix_res -> SuSiE-rRSS -SER
ser_effect <- list(
  GTEx = coloc_pooled_plot_data(load_data("Fig5A.GTEx_Coloc_Replication_all_methods"),
                                levels = coloc_method_levels_supp, labels = coloc_method_labels_supp),
  UKB  = coloc_pooled_plot_data(load_data("Fig5B.UKB_Coloc_Replication_all_methods"),
                                levels = coloc_method_levels_supp, labels = coloc_method_labels_supp)
)
map(ser_effect, ~ .x %>% filter(metric == "Non-replication rate") %>% select(method, est, lo, hi))

save_data(ser_effect, "FigS11E.Coloc_SER_Fallback_Effect")
```
