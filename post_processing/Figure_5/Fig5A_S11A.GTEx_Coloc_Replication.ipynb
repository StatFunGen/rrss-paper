# Fig 5A / S11A — GTEx eQTL colocalization replication (data processing)

Credible-set replication / non-replication rates across 47 GTEx tissues, pooled over gene × tissue pairs, with bootstrap 95% CIs. SuSiE-rRSS = EB-mix with SER fallback (`eb+SER`). Helpers in `R/utils.R` (`read_coloc_records`, `coloc_pooled_summary`, `coloc_pooled_plot_data`).

**Outputs:**
- `figure_data/Fig5A.GTEx_Coloc_Replication.rds` — ADSP reference (plot-ready)
- `figure_data/FigS11A.GTEx_Coloc_Replication_1kG.rds` — 1000 Genomes reference (plot-ready)
- `*_all_methods.{rds,csv}` — every method incl. `rss+SER` and EB-mix without SER fallback

```r
source("../../R/utils.R")
```

## Fig 5A: ADSP reference

```r
per_record  <- read_coloc_records("/sc/arion/projects/ad-omics/haochen/rss_qc/2026_output/Gtex_colocalization/coloc/result")
pooled_wide <- coloc_pooled_summary(per_record, B = 1000, seed = 2026)

save_data(pooled_wide, "Fig5A.GTEx_Coloc_Replication_all_methods")
save_data(coloc_pooled_plot_data(pooled_wide), "Fig5A.GTEx_Coloc_Replication")
```
